```python
import pandas as pd
from sklearn import datasets
import matplotlib.pyplot as plt
```

```python
# Load the dataset
df = pd.read_excel('aggregati_1.xlsx')

# Display the first few rows of the dataframe
display(df.head())
```

```python
print(df.columns)
```

```python
# Dictionary to translate column names
translation_dict = {
    'Unnamed: 0': 'Index',
    'ANNO': 'Year',
    'NUMERO': 'Delivery_Number',
    'VOLUME': 'Volume',
    'PESO': 'Weight',
    'COMMITTENTE': 'Client',
    'LINEA': 'Delivery_Line',
    'QUANTITA_IMBALLI': 'Package_Quantity',
    'VIAGGIO_RITIRO_NUMERO': 'Pickup_Trip_Number',
    'CODICE_DESTINATARIO': 'Recipient_Code',
    'RAGIONE_SOCIALE_destinatario': 'Recipient_Company_Name',
    'INDIRIZZO_destinatario': 'Recipient_Address',
    'CAP_destinatario': 'Recipient_Zip_Code',
    'LOCALITA_destinatario': 'Recipient_City',
    'PROVINCIA_destinatario': 'Recipient_Province',
    'NAZIONE_destinatario': 'Recipient_Country',
    'lat_destinatario': 'Recipient_Latitude',
    'lon_destinatario': 'Recipient_Longitude',
    'CODICE MITTENTE': 'Sender_Code',
    'Ragione_Sociale_MITTENTE': 'Sender_Company_Name',
    'INDIRIZZO_mittente': 'Sender_Address',
    'CAP_mittente': 'Sender_Zip_Code',
    'LOCALITA_mittente': 'Sender_City',
    'PROVINCIA_mittente': 'Sender_Province',
    'NAZIONE_mittente': 'Sender_Country',
    'lat_mittente': 'Sender_Latitude',
    'lon_mittente': 'Sender_Longitude',
    'QUANTITA': 'Quantity'
}

# Rename columns
df.rename(columns=translation_dict, inplace=True)

# Display the first few rows of the dataframe to verify the changes
display(df.head())
```

```python
df.drop(columns=['Quantity'], inplace=True)

# Display the first few rows of the dataframe to verify the changes
display(df.head())
```

```python
from sklearn.model_selection import train_test_split

# Split the dataset into train (60%), evaluation (20%), and test (20%) sets
train_df, temp_df = train_test_split(df, test_size=0.4, random_state=42)
eval_df, test_df = train_test_split(temp_df, test_size=0.5, random_state=42)

# Define the features (X) and target (y) for each set
X_train = train_df[['Volume']]
y_train = train_df['Weight']
X_eval = eval_df[['Volume']]
y_eval = eval_df['Weight']
X_test = test_df[['Volume']]
y_test = test_df['Weight']

# Display the number of rows in each set
print(f"Train set: {train_df.shape[0]} rows")
print(f"Evaluation set: {eval_df.shape[0]} rows")
print(f"Test set: {test_df.shape[0]} rows")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Define the features (X) and target (y)
X = df[['Volume']]  # Assuming 'Volume' is the feature for regression
y = df['Weight']

# Create and train the model
model = LinearRegression()
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)

# Calculate and print the mean squared error
mse = mean_squared_error(y_test, y_pred)
print(f"Mean Squared Error: {mse}")

# Plot the results
plt.scatter(X_test, y_test, color='black', label='Actual')
plt.plot(X_test, y_pred, color='blue', linewidth=3, label='Predicted')
plt.xlabel('Volume')
plt.ylabel('Weight')
plt.legend()
plt.show()
```
